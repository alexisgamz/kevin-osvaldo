```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo
sns.set_theme(style="whitegrid")

# --- Generación de los Conjuntos de Datos (KEVIN GAMEZ) ---

# Set 1
cats = ['Norte', 'Sur', 'Este', 'Oeste']
conteos = np.array([120, 90, 150, 110])

# Set 2
n_2 = 120
x_2 = np.linspace(0, 100, n_2)
y_2 = 2.0 * x_2 + np.random.normal(0, 25, n_2)

# Set 3
muestra = np.random.normal(loc=50, scale=10, size=1000) + np.random.exponential(scale=2, size=1000)*0.5

# Set 4
t_4 = np.arange(1, 25)
tendencia = 5 + 1.2 * t_4
estacional = 4 * np.sin(2 * np.pi * t_4 / 12)
ruido_4 = np.random.normal(0, 2, size=t_4.shape)
serie = tendencia + estacional + ruido_4

# Set 5
df = pd.DataFrame({
    'ingresos': np.random.normal(50, 10, 200),
    'gasto_marketing': np.random.normal(25, 5, 200),
    'clientes_nuevos': np.random.poisson(12, 200) + np.random.normal(0, 1, 200),
    'satisfaccion': np.clip(np.random.normal(3.5, 0.7, 200), 1, 5)
})
df['ventas'] = 2.5*df['ingresos'] + 1.1*df['gasto_marketing'] + np.random.normal(0, 8, 200)

# Set 6
categorias_6 = ['A', 'B', 'C', 'D']
valores_6 = np.array([35, 25, 22, 18])

# Set 7
fs = 500
T  = 2.0
t_7 = np.linspace(0, T, int(fs*T), endpoint=False)
f1, f2 = 50, 120
signal = 1.0*np.sin(2*np.pi*f1*t_7) + 0.6*np.sin(2*np.pi*f2*t_7) + 0.3*np.random.normal(0,1,len(t_7))
Y = np.fft.rfft(signal)
freqs = np.fft.rfftfreq(len(t_7), d=1/fs)
amplitud = np.abs(Y) / len(t_7)

# Set 8
n_8 = 60
X_8 = np.random.uniform(10, 100, size=n_8)
Y_8 = 1.5 * X_8 + np.random.normal(0, 15, size=n_8)
S_8 = np.random.uniform(50, 1500, size=n_8)

# --- Visualizaciones Recomendadas ---

fig, axes = plt.subplots(4, 2, figsize=(15, 20))
plt.subplots_adjust(hspace=0.4)

# 1. Gráfico de Barras (Set 1)
axes[0, 0].bar(cats, conteos, color='skyblue')
axes[0, 0].set_title('Set 1: Gráfico de Barras (Comparación de Categorías)')

# 2. Scatter Plot (Set 2)
axes[0, 1].scatter(x_2, y_2, alpha=0.6, color='green')
axes[0, 1].set_title('Set 2: Gráfico de Dispersión (Correlación)')

# 3. Histograma (Set 3)
sns.histplot(muestra, kde=True, ax=axes[1, 0], color='purple')
axes[1, 0].set_title('Set 3: Histograma con KDE (Distribución)')

# 4. Gráfico de Líneas (Set 4)
axes[1, 1].plot(t_4, serie, marker='o', linestyle='-', color='red')
axes[1, 1].set_title('Set 4: Gráfico de Líneas (Serie de Tiempo)')

# 5. Heatmap de Correlación (Set 5)
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', ax=axes[2, 0])
axes[2, 0].set_title('Set 5: Matriz de Correlación (Análisis Multivariado)')

# 6. Gráfico de Pastel (Set 6)
axes[2, 1].pie(valores_6, labels=categorias_6, autopct='%1.1f%%', startangle=140)
axes[2, 1].set_title('Set 6: Gráfico de Pastel (Composición)')

# 7. Gráfico de Espectro (Set 7)
axes[3, 0].plot(freqs, amplitud, color='orange')
axes[3, 0].set_title('Set 7: Análisis de Frecuencia (FFT)')
axes[3, 0].set_xlim(0, 150)

# 8. Bubble Chart (Set 8)
scatter = axes[3, 1].scatter(X_8, Y_8, s=S_8, alpha=0.5, c=X_8, cmap='viridis')
axes[3, 1].set_title('Set 8: Gráfico de Burbujas (3 Variables)')

plt.show()
```

Lo que se trabajó en este Notebook
Para esta evaluación, el enfoque no fue solo "dibujar gráficas", sino aplicar un criterio de ingeniería para decidir cómo transformar datos crudos en información útil. Cada elección se tomó pensando en qué es lo que realmente queremos descubrir en cada set:

Comparativa de Categorías (Set 1 y 6): Cuando tenemos grupos definidos como zonas geográficas o categorías de productos, lo más directo es usar barras para ver quién es quién, o un gráfico de pastel si lo que nos importa es ver cómo se reparten el "pastel" total de los datos.

Buscando Patrones y Tendencias (Set 2 y 4): Para entender si dos variables crecen juntas, el gráfico de dispersión es nuestro mejor aliado. Pero, cuando el tiempo entra en la ecuación (como en el Set 4), conectamos los puntos con una línea para exponer claramente los ciclos y las tendencias a largo plazo.

Entendiendo el "Comportamiento" (Set 3): Con mil datos encima, no sirve ver puntos individuales. Usamos un histograma para ver dónde se amontona la mayoría y le sumamos una curva de densidad (KDE) para suavizar el ruido y entender la "personalidad" de la muestra.
Visión Estratégica (Set 5 y 8): En IA, necesitamos ver el panorama completo. El Mapa de Calor (Heatmap) nos dice de un vistazo qué variables están conectadas (vital para seleccionar características en un modelo), mientras que el Gráfico de Burbujas nos permite meter una tercera dimensión (el tamaño) sin necesidad de un gráfico 3D complicado.

Extracción de Señales (Set 7): Aquí aplicamos procesamiento puro. Al pasar la señal al dominio de la frecuencia con una FFT, limpiamos el ruido para detectar los picos exactos que definen el sistema.
