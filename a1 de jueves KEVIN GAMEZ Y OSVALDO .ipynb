3 CONDICIONALES

```python
# --- 3. CONDICIONALES ---

edad = 18
tiene_entrada = True

# Ejemplo de decisión compuesta
if edad >= 18 and tiene_entrada:
    print("¡Bienvenido al concierto!")
elif edad >= 18 and not tiene_entrada:
    print("Eres mayor de edad, pero necesitas comprar un ticket.")
else:
    print("Lo siento, eres menor de edad y no puedes pasar.")

# --- CONDICIONAL EN UNA SOLA LÍNEA (Ternario) ---
# Muy útil para decisiones rápidas
estado = "Permitido" if edad >= 18 else "Denegado"
print(f"Acceso: {estado}")
```

4. BUCLES

```python
# --- 4. BUCLES (LOOPS) ---

# 1. BUCLE FOR (Ideal para recorrer colecciones o rangos fijos)
frutas = ["manzana", "pera", "uva"]

print("Lista de frutas:")
for fruta in frutas:
    print(f"- {fruta}")

# Usando range() para repetir algo un número específico de veces
print("\nContando del 1 al 5:")
for i in range(1, 6):
    print(i)


# 2. BUCLE WHILE (Se repite mientras una condición sea verdadera)
contador = 5
print("\nCuenta regresiva:")
while contador > 0:
    print(contador)
    contador -= 1  # ¡Importante! Si no restamos, el bucle sería infinito
print("¡Despegue! 🚀")


# 3. CONTROL DE BUCLES (break y continue)
print("\nEjemplo con interrupción:")
for num in range(1, 10):
    if num == 5:
        break  # Se detiene por completo en el 5
    print(num)
```

# --- 5. FUNCIONES ---

# 1. Definición de una función simple
def saludar(nombre):
    """Esta función da la bienvenida al usuario"""
    return f"Hola {nombre}, ¡bienvenido al sistema!"

# 2. Función con parámetros por defecto y múltiples valores
def calcular_total(precio, impuesto=0.16):
    total = precio + (precio * impuesto)
    return total

# 3. Llamando a las funciones (Ejecución)
mensaje = saludar("Alex")
print(mensaje)

precio_final = calcular_total(100) # Usa el 0.16 por defecto
print(f"Total con IVA: ${precio_final}")

precio_rebajado = calcular_total(100, 0.08) # Usamos un impuesto diferente
print(f"Total con impuesto reducido: ${precio_rebajado}")

6. PROGRAMACION ORIENTADA A OBJETOS

```python
# --- 6. PROGRAMACIÓN ORIENTADA A OBJETOS (POO) ---

class Coche:
    # El "Constructor" (Define las propiedades iniciales)
    def __init__(self, marca, modelo, combustible):
        self.marca = marca          # Atributo
        self.modelo = modelo        # Atributo
        self.combustible = combustible
        self.encendido = False      # Atributo por defecto

    # Métodos (Lo que el objeto puede HACER)
    def arrancar(self):
        if not self.encendido:
            self.encendido = True
            return f"El {self.marca} {self.modelo} ha arrancado. ¡Brum brum!"
        return "El coche ya estaba encendido."

    def mostrar_info(self):
        estado = "encendido" if self.encendido else "apagado"
        return f"Coche: {self.marca} {self.modelo} | Estado: {estado}"

# --- CREANDO OBJETOS (Instancias) ---
mi_coche = Coche("Toyota", "Corolla", "Gasolina")
otro_coche = Coche("Tesla", "Model 3", "Eléctrico")

# Usando los objetos
print(mi_coche.mostrar_info())
print(mi_coche.arrancar())
print(mi_coche.mostrar_info())
```
